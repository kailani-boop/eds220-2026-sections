```python
# Writing it all in one go
clean = (df.loc[df['Conflict Present'] == 'Y', 'State']
        .str.split('; ').explode()
        .unique
        )
```

## Visualize Data Exploration

```python
import matplotlib.pyplot as plt

state_counts = clean.value_counts(dropna = False)
```

```python
plt.bar(
    state_counts.index.astype(str), 
    state_counts.values,  
    color = 'skyblue', 
    edgecolor = 'black'
)
plt.xlabel('State Abbr')
plt.ylabel('Count')
plt.title('Water Conflicts Present Per State')
plt.show
```
